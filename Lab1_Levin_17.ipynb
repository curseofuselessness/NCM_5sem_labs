Даны два натуральных числа. Вывести наименьшее общее
кратное этих чисел.

```python
import numpy as np
```

```python
a = 4
b = 3
```

```python
def gcd(a, b): # НОД
    while b != 0:
        a, b = b, a % b
    return a

def lcm(a, b): # НОК
    return abs(a * b) // gcd(a, b)
```

```python
lcm(a, b)
```

Сформируйте случайным образом список чисел. Получите
новый список из положительных элементов исходного списка, стоящих на
четных местах.

```python
import random
```

```python
ARR_LENGTH = 10
```

```python
random_arr = np.random.random(ARR_LENGTH) * 2 - 1 # т.к.  разброс [0; 1], то после *2-1 будет [-1; 1]
random_arr
```

```python
ans = list(filter(lambda x: x>0, random_arr[::2]))
ans
```
